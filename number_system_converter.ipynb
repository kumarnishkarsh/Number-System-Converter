```python
def convert(value, from_base, to_base):
    decimal = int(value, from_base)
    if to_base == 10:
        return str(decimal)
    elif to_base == 2:
        return bin(decimal).replace("0b", "")
    elif to_base == 8:
        return oct(decimal).replace("0o", "")
    elif to_base == 16:
        return hex(decimal).replace("0x", "").upper()

print("Number System Converter\n")
print("Bases: 2 = Binary | 8 = Octal | 10 = Decimal | 16 = Hexadecimal\n")

value = input("Enter the number: ")
from_base = int(input("Enter current base (2/8/10/16): "))
to_base = int(input("Enter new base (2/8/10/16): "))

result = convert(value, from_base, to_base)
print(f"\nResult: {value} (base {from_base}) = {result} (base {to_base})")
```
